```python
# Este codigo lo paso el profesor para que lo peguemos

import matplotlib.pyplot as plt
import numpy as np

plt.style.use('dark_background')
fig = plt.figure(figsize=(14, 6))

# --- Gráfico 1: función seno con relleno y anotación ---
ax1 = fig.add_subplot(1, 2, 1)
x = np.linspace(0, 4 * np.pi, 400)
y = np.sin(x) * np.exp(-x / 10)

ax1.plot(x, y, color='cyan', linewidth=2)
ax1.fill_between(x, y, alpha=0.3, color='cyan')
ax1.set_title('Oscilación amortiguada', fontsize=14)
ax1.annotate('Pico máximo', xy=(x[np.argmax(y)], max(y)),
             xytext=(10, 0.5), arrowprops=dict(arrowstyle='->', color='white'))
ax1.grid(alpha=0.2)

# --- Gráfico 2: superficie 3D ---
ax2 = fig.add_subplot(1, 2, 2, projection='3d')
X = np.linspace(-5, 5, 100)
Y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(X, Y)
Z = np.sin(np.sqrt(X**2 + Y**2))

ax2.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')
ax2.set_title('Superficie 3D: sin(√(x²+y²))', fontsize=14)

plt.tight_layout()
plt.show()
```

```python
# Lo que me genero a mi Claude
import matplotlib.pyplot as plt
import numpy as np

# Datos de ejemplo
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

categorias = ['A', 'B', 'C', 'D', 'E']
valores = [23, 45, 12, 67, 34]

theta = np.linspace(0, 2 * np.pi, 50)
r = np.random.rand(50) * 2

datos_dispersión = np.random.randn(200, 2)

# Crear figura con 4 subplots (2x2)
fig, axs = plt.subplots(2, 2, figsize=(12, 10))
fig.suptitle('Ejemplos con Matplotlib', fontsize=16, fontweight='bold')

# 1. Gráfico de líneas
axs[0, 0].plot(x, y1, label='seno', color='royalblue', linewidth=2)
axs[0, 0].plot(x, y2, label='coseno', color='crimson', linewidth=2, linestyle='--')
axs[0, 0].set_title('Gráfico de líneas')
axs[0, 0].legend()
axs[0, 0].grid(True, alpha=0.3)

# 2. Gráfico de barras
axs[0, 1].bar(categorias, valores, color='mediumseagreen')
axs[0, 1].set_title('Gráfico de barras')
axs[0, 1].set_ylabel('Valores')

# 3. Gráfico de dispersión (scatter)
axs[1, 0].scatter(datos_dispersión[:, 0], datos_dispersión[:, 1],
                   c=datos_dispersión[:, 0], cmap='viridis', alpha=0.7)
axs[1, 0].set_title('Gráfico de dispersión')

# 4. Gráfico polar (dentro de un subplot normal, usando proyección)
axs[1, 1].remove()
ax_polar = fig.add_subplot(2, 2, 4, projection='polar')
ax_polar.plot(theta, r, color='darkorange')
ax_polar.set_title('Gráfico polar')

plt.tight_layout()
plt.show()
```
