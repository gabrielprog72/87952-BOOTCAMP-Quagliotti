```python
from math import gcd

def simplificar(numerador, denominador):
    if denominador == 0:
        raise ValueError("El denominador no puede ser 0")
    divisor = gcd(numerador, denominador)
    numerador //= divisor
    denominador //= divisor
    # Mantener el signo negativo siempre en el numerador
    if denominador < 0:
        numerador = -numerador
        denominador = -denominador
    return numerador, denominador


def pedir_fraccion(mensaje):
    print(mensaje)
    while True:
        try:
            num = int(input("  Numerador: "))
            den = int(input("  Denominador: "))
            if den == 0:
                print("  El denominador no puede ser 0. Intenta de nuevo.")
                continue
            return num, den
        except ValueError:
            print("  Por favor ingresa números enteros válidos.")


def sumar_fracciones(n1, d1, n2, d2):
    num = n1 * d2 + n2 * d1
    den = d1 * d2
    return simplificar(num, den)


def multiplicar_fracciones(n1, d1, n2, d2):
    num = n1 * n2
    den = d1 * d2
    return simplificar(num, den)


def main():
    print("=== Suma y multiplicación de dos fracciones ===\n")

    n1, d1 = pedir_fraccion("Primera fracción:")
    n2, d2 = pedir_fraccion("\nSegunda fracción:")

    suma_num, suma_den = sumar_fracciones(n1, d1, n2, d2)
    mult_num, mult_den = multiplicar_fracciones(n1, d1, n2, d2)

    print("\n--- Resultados ---")
    print(f"Suma:          {n1}/{d1} + {n2}/{d2} = {suma_num}/{suma_den}")
    print(f"Multiplicación: {n1}/{d1} * {n2}/{d2} = {mult_num}/{mult_den}")


if __name__ == "__main__":
    main()
```
