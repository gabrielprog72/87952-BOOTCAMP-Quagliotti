```python
from math import gcd

def simplificar(numerador, denominador):
    if denominador == 0:
        raise ValueError("El denominador no puede ser 0")
    divisor = gcd(numerador, denominador)
    numerador //= divisor
    denominador //= divisor
    # Mantener el signo negativo siempre en el numerador
    if denominador < 0:
        numerador = -numerador
        denominador = -denominador
    return numerador, denominador


def pedir_fraccion(mensaje):
    print(mensaje)
    while True:
        try:
            num = int(input("  Numerador: "))
            den = int(input("  Denominador: "))
            if den == 0:
                print("  El denominador no puede ser 0. Intenta de nuevo.")
                continue
            return num, den
        except ValueError:
            print("  Por favor ingresa números enteros válidos.")


def sumar_fracciones(n1, d1, n2, d2):
    num = n1 * d2 + n2 * d1
    den = d1 * d2
    return simplificar(num, den)


def multiplicar_fracciones(n1, d1, n2, d2):
    num = n1 * n2
    den = d1 * d2
    return simplificar(num, den)


def main():
    print("=== Suma y multiplicación de dos fracciones ===\n")

    n1, d1 = pedir_fraccion("Primera fracción:")
    n2, d2 = pedir_fraccion("\nSegunda fracción:")

    suma_num, suma_den = sumar_fracciones(n1, d1, n2, d2)
    mult_num, mult_den = multiplicar_fracciones(n1, d1, n2, d2)

    print("\n--- Resultados ---")
    print(f"Suma:          {n1}/{d1} + {n2}/{d2} = {suma_num}/{suma_den}")
    print(f"Multiplicación: {n1}/{d1} * {n2}/{d2} = {mult_num}/{mult_den}")


if __name__ == "__main__":
    main()
```

```python
from math import gcd


class Fraccion:
    """Representa una fracción y permite operar con otras fracciones."""

    def __init__(self, numerador, denominador):
        if denominador == 0:
            raise ZeroDivisionError("El denominador no puede ser 0")
        self.numerador = numerador
        self.denominador = denominador

    def simplificar(self):
        divisor = gcd(self.numerador, self.denominador)
        num = self.numerador // divisor
        den = self.denominador // divisor
        # Mantener el signo negativo siempre en el numerador
        if den < 0:
            num = -num
            den = -den
        return Fraccion(num, den)

    def sumar(self, otra):
        num = self.numerador * otra.denominador + otra.numerador * self.denominador
        den = self.denominador * otra.denominador
        return Fraccion(num, den).simplificar()

    def multiplicar(self, otra):
        num = self.numerador * otra.numerador
        den = self.denominador * otra.denominador
        return Fraccion(num, den).simplificar()

    def __str__(self):
        return f"{self.numerador}/{self.denominador}"


class LectorDeFraccion:
    """Se encarga de pedir al usuario los datos de una fracción, validando el denominador."""

    @staticmethod
    def pedir(mensaje):
        print(mensaje)
        while True:
            try:
                num = int(input("  Numerador: "))
                den = int(input("  Denominador: "))
                return Fraccion(num, den)
            except ZeroDivisionError:
                print("  El denominador no puede ser 0. Intenta de nuevo.")
            except ValueError:
                print("  Por favor ingresa números enteros válidos.")


class CalculadoraDeFracciones:
    """Orquesta el flujo del programa: pide las fracciones y muestra los resultados."""

    def ejecutar(self):
        print("=== Suma y multiplicación de dos fracciones (POO) ===\n")

        f1 = LectorDeFraccion.pedir("Primera fracción:")
        f2 = LectorDeFraccion.pedir("\nSegunda fracción:")

        suma = f1.sumar(f2)
        multiplicacion = f1.multiplicar(f2)

        print("\n--- Resultados ---")
        print(f"Suma:           {f1} + {f2} = {suma}")
        print(f"Multiplicación: {f1} * {f2} = {multiplicacion}")


if __name__ == "__main__":
    CalculadoraDeFracciones().ejecutar()
```
